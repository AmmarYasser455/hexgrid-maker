# hexgrid-maker demo

A 500 m hexagonal grid over five Paris landmarks.

```python
import importlib.util
import os
import subprocess
import sys

if importlib.util.find_spec('hexgrid_maker') is None:  # fresh runtime such as Colab: install first
    subprocess.check_call([sys.executable, '-m', 'pip', 'install', '-q', 'git+https://github.com/AmmarYasser455/hexgrid-maker'])
from hexgrid_maker import hex_grid, load_geometry

path = 'examples/paris_landmarks.geojson'
if not os.path.exists(path):  # e.g. Colab: fetch the sample file
    import urllib.request
    path = 'paris_landmarks.geojson'
    urllib.request.urlretrieve('https://raw.githubusercontent.com/AmmarYasser455/hexgrid-maker/main/examples/paris_landmarks.geojson', path)
hull = load_geometry(path)
print(hull.geom_type, [round(v, 4) for v in hull.bounds])
```

```python
grid = hex_grid(hull, 500, orientation='flat')
print(len(grid['features']), 'cells')
print(grid['features'][0]['properties'])
```

```python
from shapely.geometry import Point, shape

eiffel = Point(2.2945, 48.8584)
cells = [f for f in grid['features'] if shape(f['geometry']).contains(eiffel)]
print('cells containing the Eiffel Tower:', len(cells))
assert len(cells) == 1
```
